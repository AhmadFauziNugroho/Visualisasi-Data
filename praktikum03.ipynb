**Import Liblary**

```python
import matplotlib.pyplot as plt
```

**Dataset**

```python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'mei', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
product_A_sales = [10,20,15,18,20,22,30,25,40,45,23,11]
product_B_sales = [15,25,20,7,12,45,12,14,55,23,45,66]
```

**Single Line Plot**

```python
plt.plot(months, product_A_sales, label='Product A')
plt.title('Penjualan Produk A per Bulan')
plt.xlabel('Bulan')
plt.ylabel('Penjualan')
plt.legend()
plt.grid(True)
plt.show()
```

**Multiple Line & Kustomization**

```python
plt.plot(months, product_A_sales, label='Product A', linestyle="--", marker='o')
plt.plot(months, product_B_sales, label='Product B', linestyle="-", marker='x')
plt.title('Penjualan Produk per Bulan')
plt.xlabel('Bulan')
plt.ylabel('Penjualan')
plt.legend()
plt.grid(True)
plt.show()
```

**Jenis Garis untuk Menunjukan Tren**

```python
product_C_sales = [30,12,22,14,12,15,23,67,44,66,11,22]
product_D_sales = [10,20,1,25,77,55,33,53,72,34,64,12]
```

```python
plt.plot(months, product_A_sales, label='Product A', linestyle="--", color='red')
plt.plot(months, product_B_sales, label='Product B', linestyle="-", color='blue')
plt.plot(months, product_C_sales, label='Product C', linestyle=":", color='purple')
plt.plot(months, product_D_sales, label='Product D', linestyle="-.", color='green')
plt.title('Penjualan Produk per Bulan')
plt.xlabel('Bulan')
plt.ylabel('Jumlah Penjualan')
plt.legend()
plt.grid(True)
plt.show()
```

**Subplot pada Line Chart**

```python
fig, axs = plt.subplots(2, 1, figsize=(10, 8))

axs[0].plot(months, product_C_sales, label = 'product C', color ='green', marker='d')
axs[0].set_title('Penjualan Product C Per Bulan')
axs[0].set_xlabel('Bulan')
axs[0].set_ylabel('Jumlah Penjualan')
axs[0].legend()
axs[0].grid(True)

axs[1].plot(months, product_D_sales, label = 'product D', color ='orange', marker='s')
axs[1].set_title('Penjualan Product D Per Bulan')
axs[1].set_xlabel('Bulan')
axs[1].set_ylabel('Jumlah Penjualan')
axs[1].legend()
axs[1].grid(True)

plt.tight_layout()
plt.show()
```
