**Import Liblary**

```python
import matplotlib.pyplot as plt
```

**Dataset**

```python
suhu = [20, 21, 22, 23, 24, 25, 26, 27, 28]
penjualan = [100, 200, 300, 400, 500, 600, 700, 800, 900]
```

**1. Basic Scatter Plot**

```python
plt.scatter(suhu, penjualan)
plt.title('hubungan penjualan es krim dengan suhu')
plt.xlabel('Suhu')
plt.ylabel('Penjualan')
plt.show()
```

**2. Kustomisasi Scatter Plot**

```python
plt.scatter(suhu, penjualan, color='red', s=100, edgecolors='black', alpha=0.7)
plt.title('hubungan penjualan es krim dengan suhu')
plt.xlabel('Suhu')
plt.ylabel('Penjualan')
plt.grid(True)
plt.show()
```

**3. Multiple Scatter Plot**

```python
suhu = [20, 21, 22, 23, 24, 25, 26, 27, 28]

penjualan_weekdays = [50, 60, 70, 80, 90, 100, 110, 120, 130]
penjualan_weekends = [70, 80, 90, 100, 110, 120, 130, 140, 150]

plt.scatter(suhu, penjualan_weekdays, color="blue", label="Weekdays", s=80)
plt.scatter(suhu, penjualan_weekends, color="red", label="Weekends", s=80)

plt.title("hubungan penjualan es krim dengan suhu")
plt.xlabel("Suhu")
plt.ylabel("Penjualan")
plt.legend()
plt.show()
```

**4. Hubungan Antar Variabel**

```python
suhu = [20, 21, 22, 23, 24, 25, 26, 27, 28]
penjualan = [100, 200, 300, 400, 500, 600, 700, 800, 900]

kelembaban = [40, 45, 50, 55, 60, 65, 70, 75, 80]
ukuran_titik = [150, 155, 160, 165, 170, 175, 180, 185, 190]

plt.scatter(suhu, penjualan, c=kelembaban, s=ukuran_titik, cmap='coolwarm', alpha=0.6)
plt.title('hubungan penjualan, suhu, dan kelembaban')
plt.xlabel('Suhu')
plt.ylabel('Penjualan Es Krim')
plt.colorbar(label='Kelembaban (%)')
plt.show()
```

**5. 3d Scatter Plot**

```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

suhu = [20, 21, 22, 23, 24, 25, 26, 27, 28]

penjualan_weekdays = [50, 60, 70, 80, 90, 100, 110, 120, 130]
penjualan_weekends = [70, 80, 90, 100, 110, 120, 130, 140, 150]

fig = plt.figure(figsize=(8, 10))
ax = fig.add_subplot(111, projection='3d')

ax.scatter(suhu, penjualan_weekdays, penjualan_weekends, color='darkblue', s=80, label='Data Penjualan')

ax.set_title('hubungan penjualan Es Krim dan Suhu (3D)')
ax.set_xlabel('Suhu')
ax.set_ylabel('Penjualan Weekdays')
ax.set_zlabel('Penjualan Weekends')
ax.legend()
plt.show()
```
